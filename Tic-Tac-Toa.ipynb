```python
#importing Packages from tkinter
from tkinter import *
from tkinter import messagebox

Player1 = 'X'
stop_game = False

########## start of hannane's code - PART 2 ##############

def clicked(r,c):

    #player
    global Player1
    # global stop_game

    if Player1 == "X" and states[r][c] == 0 and stop_game == False:
        b[r][c].configure(text = "X")
        states[r][c] = 'X'
        Player1='O'

    if Player1 == 'O' and states[r][c] == 0 and stop_game == False:
        b[r][c].configure(text = 'O')
        states[r][c] = "O"
        Player1 = "X"

    check_if_win()

############ end of hannane's code - PART 2 ##############


########## PART 3 - CHECKING THE WINNER ##############

#change the color of winner box
def show_winner(cells, mark):
    global stop_game
    stop_game = True
    for r, c in cells:
        b[r][c].configure(
            #bottom background color
            bg="lightgreen",
            #background of bottom when clicking it
            activebackground="lightgreen"
        )

    messagebox.showinfo("Winner", f"{mark} is the winner.")

#check who is the winner of the game
def check_if_win():

    #read argument outside the function
    global stop_game

    #creat a for loop to check the status of database here is 'state' matrix
    for i in range(3):

        #checking the boxes horizontaly
        if states[i][0] == states[i][1] == states[i][2] !=0:
            show_winner([(i, 0), (i, 1), (i, 2)], states[i][0])
            return

        #checking the boxes vertically
        elif states [0][i] == states[1][i] == states[2][i] != 0:
            show_winner([(0, i), (1, i), (2, i)], states[i][0])
            return

        #checking the boxes diagonal left to right
        elif states[0][0] == states[1][1] == states[2][2] !=0:
            show_winner([(0, 0), (1, 1), (2, 2)], states[i][0])
            return

        #checking the boxes diagonal right to lef
        elif states[0][2] == states[1][1] == states[2][0] !=0:
            show_winner([(0, 2), (1, 1), (2, 0)], states[i][0])
            return

        # ensuring all boxes are full fill and the game status is tie
        elif states[0][0] and states[0][1] and states[0][2] and states[1][0] and states[1][1] and states[1][2] and states[2][0] and states[2][1] and states[2][2] != 0:
            stop_game = True
            winner = messagebox.showinfo("win-win", "the score is tied")

############ END OF PART 3 ############

############## part 1: Maryam ##############
# creating window
window = Tk()

# Title of the window
window.title("Pathway-TicTacToe")
window.resizable(10,10)

#Button
b = [
     [0,0,0],
     [0,0,0],
     [0,0,0]]

#Data Base
states = [
     [0,0,0],
     [0,0,0],
     [0,0,0]]

#creat 3*3 botton as playground
for i in range(3):
    for j in range(3):

        b[i][j] = Button(
                        height = 4, width = 8,
                        font = ("Arial","20"),
                        #call click function when user click any botton
                        command = lambda r = i, c = j : clicked(r,c))
        #placing botton on windows
        b[i][j].grid(row = i, column = j)

# keep the windows open
mainloop()
```
